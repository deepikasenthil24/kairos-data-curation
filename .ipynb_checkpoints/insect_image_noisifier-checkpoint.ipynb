# Insect Image Noisifier
Compatible for ResNet and input to embedding generator for Kairos

```python
import os
import torch
from PIL import Image
from tqdm import tqdm
import numpy as np
from datasets import load_dataset

import pandas as pd
from utils.label_mappings import *

OUT_DIR = 'data'
os.makedirs(OUT_DIR, exist_ok=True)# load iNat data


iNat36 = load_dataset("sxj1215/inaturalist", split='train') #36k rows #3.3 GB
ids = list(range(len(iNat36)))
iNat36 = iNat36.add_column("id", ids) #not idempotent

iNat36_label_df = pd.DataFrame({'messages': iNat36['messages'], 'id': iNat36['id']})

def get_iNat_label(messages):
    return messages[1]['content']
    
iNat36_label_df['species'] = iNat36_label_df['messages'].apply(get_iNat_label)

def map_inat_to_clean_label(label):
    if label in iNat_to_clean_map:
        return iNat_to_clean_map[label]
    else:
        return 'non_insect'
        #add index to noisy index list here?
        
iNat36_label_df['clean_label'] = iNat36_label_df['species'].apply(map_inat_to_clean_label)
print(iNat36_label_df['clean_label'].unique()) # check that we are using the correct classes
```

```python
iNat36_label_df[iNat36_label_df['clean_label'] != 'non_insect'].iloc[0]
```

```python
insect_mask = iNat36_label_df['clean_label'] != 'non_insect'
insect_dataset = iNat36.select(np.where(insect_mask)[0])
```

```python
insect_dataset_labels = list(iNat36_label_df[insect_mask]['clean_label'])
```

```python
from PIL import Image
import numpy as np

inat_insect_images = []
shapes_set = set()
for img_list in insect_dataset["images"]:
    img = img_list[0]          # PIL image
    img = np.array(img)
    inat_insect_images.append(img)
    shapes_set.add(img.shape)
```

```python
print(len(inat_insect_images))
#print(shapes_set)
```

```python
import numpy as np

def add_gaussian_noise(images, noise_rate=0.4, noise_std=90.0, seed=1):
    """
    images: list of numpy arrays (H, W, 3), dtype uint8
    noise_rate: fraction of images to corrupt
    noise_std: std of Gaussian noise in pixel units (0–255 scale)
    """
    
    np.random.seed(seed)
    
    n_samples = len(images)
    n_noisy = int(n_samples * noise_rate)
    
    noisy_indices = np.random.choice(n_samples, n_noisy, replace=False)
    
    corrupted_images = []
    
    for i, img in enumerate(images):
        img_float = img.astype(np.float32)
        
        if i in noisy_indices:
            noise = np.random.normal(
                loc=0.0,
                scale=noise_std,
                size=img_float.shape
            )
            img_float = img_float + noise
        
        # clip to valid pixel range
        img_float = np.clip(img_float, 0, 255)
        
        corrupted_images.append(img_float.astype(np.uint8))

    with open(f'kairos_results/noise_{str(noise_std)[:1]}std/params_metrics.txt', 'w') as file:
        file.writelines([
            f'noise_rate: {noise_rate}\n',
            f'noise_std: {noise_std}\n',
            f'n_samples: {n_samples}\n',
            f'n_noisy: {n_noisy}\n'
        ])
        
    return corrupted_images, noisy_indices
```

```python
noisified_insect_images, noisy_insect_indexes = add_gaussian_noise(inat_insect_images)
len(noisified_insect_images)
```

```python
#save noisified images to be fed into embedding generator
save_dir = os.path.join(OUT_DIR, "noisy_images")
os.makedirs(save_dir, exist_ok=True)

for i, img in enumerate(noisified_insect_images): #takes a few minutes
    Image.fromarray(img).save(
        os.path.join(save_dir, f"img_{i}.png")
    )

np.save(os.path.join(OUT_DIR, f"noisified_inat_insect_indexes.npy"), noisy_insect_indexes)
```

```python
np.save(os.path.join(OUT_DIR, f"noisified_inat_insect_labels.npy"), insect_dataset_labels)
```
